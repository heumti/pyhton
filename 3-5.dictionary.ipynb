## 5. 딕셔너리 관련 함수

key 리스트 만들기 : keys()

```python
a = {'name' : 'pgh' , 'phone' : '010-8317=7926' ,'birth' : '1114'}
a.keys()
```

```python
type(a.keys())
```

```python
list(a.keys())
```

```python
type(list(a.keys()))
```

```python
for k in a.keys():
    print(k)
```

value 리스트 만들기 : values()

```python
a.values()
```

```python
list(a.values())
```

key, value 쌍으로 얻기 : .itmes()

```python
a.items()
```

```python
list(a.items())
```

key: value 쌍 모두 지우기 : clear()

```python
a.clear()
a
```

key값으로 벨류 얻기  : .get(key)

```python
a = {'name' : 'pgh' , 'phone' : '010-8317=7926' ,'birth' : '1114'}
a.get('name')
a.get('phone')
```

```python
print(a.get('age'))
```

```python
print(a['age'])
```

```python
# 딕셔너리 안에 찾는  key가 없으면 디폴트 값을 가져오게 할 수 있다.
# get(key,'default value)

a.get('age',25)
```

```python

a.get('age')
```
