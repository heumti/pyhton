## 10. format()함수를 사용한 포매팅

### 1) 숫자 바로 대입

```python
"I eat {} apples.".format(3)
```

```python
"I eat {} apples.".format("five")
```

```python
number =3
"I eat {} apples.".format(number) 
```

```python
number = 3 
day = "three"
"I ate {0} apples. so I was sick for {1} days.".format(number, day)
```

```python
"I ate {number} apples. so I was sick for {day} days.".format(day='three' , number = 10)
```

```python
# 인덱스와 변수명을 혼용해서 넣기 
"I ate {0} apples. so I was sick for {day} days.".format(10 , day ='three')
```

```python
# 왼쪽 정렬 
"{:<10}".format('hi')
```

```python
# 오른쪽 정렬
"{:>10}".format('hi')
```

```python
# 가운데 정렬
"{:>10}".format('hi')
```

```python
# 공백 채우기 
"{:=>10}".format('hi')
```

```python
"{:!<10}".format('hi')
```

```python
#소수점 표현
"{:0.4f}".format(3.141592)
```

```python
"{:>10.4f}".format(3.141592)
```

```python
"{{and}}".format()
```
