## 3. 변수를 만드는 여러가지 방법

튜플로 변수 만들기 .

```python
a,b = ('python','life')
print(a,b)
a,b = 'python' , 'life'
```

```python
(a,b) ='python','life'
print(a,b)
```

리스트로 변수 만들기.

```python
[a , b] =['python','life']
a
```

```python
b
```

여러 개의 변수에 같은 값을 대입

```python
a = b= c ="python"
print(a,b,c)
```

두 변수의 값 바꾸기 

```python
# 일반적인 프로그래밍에서 두 변수 값 교체 할 떄
a = 3 
b = 5
print(a,b)
```

```python
temp = a
a = b
b = temp
print(a,b)
```

```python
#python 에서
a = 3 
b = 5 
a,b = b,a
print(a,b)
```
