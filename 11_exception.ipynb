## 1. 예외 처리 기법 

try - except 구문

```python
try:
    4/0
except ZeroDivisionError as e:
    print(e)
```

```python
try:
    print(4/3)
except ZeroDivisionError as e:
    print(e)
```

try-finally ans

```python
try:
    f=open("foo.text","r")
except FileNotFoundError as e:
    print(e)
finally:
    f.close()
```

```python
try:
    f=open("foo.text" , "w")
    #예외발생 코드 
finally:
    f.close()
```

여러개의 예외 처리

```python
try:
    4/0
    a=[1,2]
    print(a[3])
    
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
except IndexError:
    print("인덱싱 할 수 없습니다.")
```

```python
try:
    a=[1,2]
    a[3]
    print(a[3])
    4/0
except Exception as e:
    print(e)
```

```python
try:
    age=int(input("나이를 입력하세요."))
except:
    print("입력이 정확하지 않습니다.")
else:
    if age <= 18:
        print("미성년자는 출입금지입니다.")
    else:
        print("환영합니다.")
```

## 3. 예외 발생시키기

```python
class Brid:
    def fly(self):
        raise NotImplementedError
```

```python
class Eagle(Brid):
    pass

```

```python
class Eagle(Brid):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

## 4. 예외 만들기 

```python
class MyError(Exception):
    def __str__(self):
        return  "허용되지 않은 별명입니다."
```

```python
def say_nick(nick):
    if nick =="바보":
        raise MyError() 
    else:
        print(nick)
```

```python
say_nick("바보")
```

```python
try:
    say_nick("천사")
    say_nick("바보")
except MyError:
    print("허용되지 않는 별명입니다.")
```

```python
try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
```
