## 3. 다양한 조건을 판단하는 elif 

```python
# if else 만으로 표현 
pocket = ['paper','cellphone']
card = True
if 'money' in pocket:
    print("택시를 타고 가라.")
else:
    if card:
        print("택시를 타고 가라.")
    else:
        print("걸어가라.")
```

```python
# elif를 사용
if 'money' in pocket :
    print("택시를 타고 가라.")
elif card:
    print("택시를 타고 가라.")
else:
    print("걸어가라")
```

```python
# if 문을 한줄로 작성하기
if 'money' in pocket : pass
else: print("카드를 꺼내라")
```
