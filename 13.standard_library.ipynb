random:난수(규칙이 없는 임의의 수)를 발생

```python
import random
random.random()
```

```python
#randint():정수값에서 난수 발생
random.randint(1,10)
```

```python
# random _pop함수
import random
random.random()
def random_pop(data):
    num=random.randint(0, len(data)-1)
    print(data,num)
    return data.pop(num)

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

```python

def random_pop(data):
    num=random.choice(data)
    print(data,num)
    data.remove(num)
    return num

data =[1,2,3,4,5]
while data:
    print(random_pop(data))
```
